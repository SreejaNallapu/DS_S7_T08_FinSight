```python
#Data inspection 
import pandas as pd
from sklearn.datasets import load_iris
# Load iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df['species'] = iris.target
# Map species numbers to names
df['species'] = df['species'].map({0:'setosa', 1:'versicolor', 2:'virginica'})
# Inspect
print(df.head())
print(df.info())                                              
print(df.isnull().sum())
```

```python
# Categorical summary
print(df.describe())
print(df['species'].value_counts()) 
```

```python
#Data Understanding

import seaborn as sns
import matplotlib.pyplot as plt
sns.pairplot(df, hue="species")
plt.show()
#Pair Plot():
sns.boxplot(x="species", y="petal length (cm)", data=df)
plt.show()
#BoxPlot():
sns.heatmap(df.drop(columns="species").corr(), annot=True, cmap="coolwarm")
plt.show()

```

```python
#IQR claculation
import seaborn as sns
import pandas as pd
import numpy as np

# Loading the dataset
df = pd.read_csv("temporal.csv")

print(df)

# Q1
Q1 = np.percentile(df['data science'], 25, method='midpoint')

# Q3
Q3 = np.percentile(df['data science'], 75, method='midpoint')

# IQR
IQR = Q3 - Q1

print("The output of IQR")
print(IQR)

# Upper bound
upper = Q3 + 1.5 * IQR

upper_array = np.array(df['data science'] >= upper)

print("Upper Bound:", upper)
print(upper_array.sum())

# Lower bound
lower = Q1 - 1.5 * IQR

lower_array = np.array(df['data science'] <= lower)

print("Lower Bound:", lower)
print(lower_array.sum())
```

```python
#violin Plot
import matplotlib.pyplot as plt
import numpy as np
# Sample datasets
cse = [80, 85, 90, 88, 92, 75, 78, 85]
aiml = [70, 75, 95, 85, 80, 72, 78, 88]
# Create violin plot
plt.violinplot(
    dataset=[cse, aiml],      # Data to plot
    positions=[1, 2],         # Position of violins on X-axis. 1- CSE, 2-AIML
    vert=True,                # Vertical violin
    widths=0.5,               # Width of violin
    showmeans=True,           # Display mean
    showextrema=True,         # Show min/max
    showmedians=True,         # Show median
    points=100                # KDE points - curve is evaluated at 100 equally spaced points along the data range
)
# Labels
plt.xticks([1, 2], ['CSE', 'AIML'])
plt.title("Violin Plot of Student Marks")
plt.xlabel("Department")
plt.ylabel("Marks")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
# Load Dataset
iris = sns.load_dataset("iris")
# Pair Plot
sns.pairplot(
    data=iris,                  # Dataset
    hue="species",              # Color by species
    hue_order=["setosa", "versicolor", "virginica"],
    palette="Set1",             # Color palette
    vars=[
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    kind="scatter",             # Scatter plots
    diag_kind="hist",           # Histogram on diagonal
    markers=["o", "s", "D"],    # Different markers
    height=3,                   # Height of each plot
    aspect=1.2,                 # Width/Height ratio
    corner=False,               # Show full matrix
    dropna=True                 # Remove missing values
)
plt.show()
```

```python
#Callbacks 
import dash
from dash import dcc, html, Input, Output

# Create Dash application
app = dash.Dash(__name__)

# Define layout
app.layout = html.Div([
    
    html.H1("Square Calculator"),

    # Input box
    dcc.Input(
        id='input',
        type='text',
        placeholder='Enter a number'
    ),

    # Output area
    html.Div(
        id='output'
    )
])

# Callback
@app.callback(
    Output(component_id='output', component_property='children'),
    Input(component_id='input', component_property='value')
)
def update_value(input_data):

    try:
        return str(float(input_data) ** 2)

    except:
        return "Error, the input is not a number"

# Run application
app.run(debug=True)
```
